```python
import pandas as pd
```

```python
chrom_data=pd.read_excel('/content/chrosome_basepairs w.xlsx')
```

```python
# Calculate the total number of base pairs across all chromosomes
total_basepairs = chrom_data['basepais(bp)'].sum()

# Convert total base pairs to Angstroms (1 bp = 3.4 Angstroms)
length_angstroms = total_basepairs * 3.4

# Convert Angstroms to meters (1 Angstrom = 1e-10 meters)
length_meters = length_angstroms * 1e-10

print(f"Total base pairs in the genome: {total_basepairs:,}")
print(f"Total length of the genome: {length_angstroms:,.2f} Angstroms")
print(f"Total length of the genome: {length_meters:.4e} meters")
```

```python
import matplotlib.pyplot as plt

# Define the years and corresponding cost per 1 Megabase (1 Mbps)
years = ['2001', '2010', '2011', '2020']
cost_per_mbps = [10000, 1.0, 0.10, 0.01]

# Total megabases in the genome (Total base pairs / 1,000,000)
total_megabases = total_basepairs / 1_000_000

# Calculate total cost for sequencing the whole genome for each year
total_costs = [cost * total_megabases for cost in cost_per_mbps]

# Print out the calculated costs for verification
for yr, cost in zip(years, total_costs):
    print(f"Year {yr}: ${cost:,.2f}")

# Plotting the cost changes over time (using a logarithmic scale due to exponential drop)
fig_cost, ax_cost = plt.subplots(figsize=(8, 5))
ax_cost.plot(years, total_costs, marker='o', color='purple', linestyle='-', linewidth=2, markersize=8)
ax_cost.set_yscale('log')  # Logarithmic scale highlights the massive drop over time
ax_cost.set_xlabel('Year', fontsize=12)
ax_cost.set_ylabel('Total Cost to Sequence Whole Genome (USD, log scale)', fontsize=12)
ax_cost.set_title('Decline in Whole Genome Sequencing Cost (2001–2020)', fontsize=14)
ax_cost.grid(True, which="both", linestyle='--', alpha=0.6)

# Annotate data points on the plot
for i, txt in enumerate(total_costs):
    ax_cost.annotate(f"${txt:,.2f}", (years[i], total_costs[i]), textcoords="offset points", xytext=(0,10), ha='center')

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt

# Define the years and corresponding calculated total costs
years = ['2001', '2010', '2011', '2020']
total_costs = [30705356.48, 3070.54, 307.05, 30.71]  # Or use your dynamic 'total_costs' variable if in the same session

# Create the bar chart
fig, ax = plt.subplots(figsize=(8, 5))
bars = ax.bar(years, total_costs, color='purple', alpha=0.8)

# Use a logarithmic scale on the y-axis to account for the dramatic drop in cost
ax.set_yscale('log')
ax.set_xlabel('Year', fontsize=12)
ax.set_ylabel('Total Cost (USD, log scale)', fontsize=12)
ax.set_title('Decline in Whole Genome Sequencing Cost (Bar Graph)', fontsize=14)
ax.grid(True, which="both", linestyle='--', alpha=0.6)

# Add text labels above each bar for clarity
for bar in bars:
    yval = bar.get_height()
    ax.text(bar.get_x() + bar.get_width()/2.0, yval, f"${yval:,.2f}",
            ha='center', va='bottom', rotation=15, fontsize=9)

plt.tight_layout()
plt.show()
```
